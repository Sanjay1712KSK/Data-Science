## Generate a NumPy identity array (4x4) and an eye array. Reshape a flat array of 12 elements into 3x4 and 4x3; perform transposition and flattening. Illustrate iterating over the reshaped array with code for environmental sensor data simulation.

```python
import numpy as np

print("===== NUMPY MATRIX OPERATIONS DEMO =====\n")
identity_matrix = np.identity(4)
eye_matrix = np.eye(4)

print("Identity Matrix (4x4):\n", identity_matrix)
print("\nEye Matrix (4x4):\n", eye_matrix)
flat_array = np.arange(1, 13)
print("\nFlat Array (12 elements):\n", flat_array)
array_3x4 = flat_array.reshape(3, 4)
array_4x3 = flat_array.reshape(4, 3)
print("\nReshaped to 3x4:\n", array_3x4)
print("\nReshaped to 4x3:\n", array_4x3)
transpose_array = array_3x4.T
print("\nTranspose of 3x4 (4x3):\n", transpose_array)
flattened_array = array_3x4.flatten()
print("\nFlattened Array:\n", flattened_array)
print("\n===== Environmental Sensor Data Simulation =====")
sensor_data = np.array([
    [32.5, 65.2, 410, 1012],
    [30.1, 70.3, 420, 1010],
    [29.8, 75.0, 415, 1008]
])

parameters = ["Temperature (°C)", "Humidity (%)", "CO2 (ppm)", "Pressure (hPa)"]
for i, location in enumerate(sensor_data):
    print(f"\nLocation {i+1}:")
    for j, value in enumerate(location):
        print(f"  {parameters[j]}: {value}")
print("\nAll Sensor Values (Element-wise using nditer):")
for value in np.nditer(sensor_data):
    print(value, end=" ")

print("\n\n===== END OF DEMO =====")
```
